CREATION OF SESSION AND CONNECTTING DATABASE AND SCHEMA


```

from snowflake.snowpark.context import get_active_session
session = get_active_session()
session.sql("USE DATABASE NOTEBOOK").collect()
session.sql("USE SCHEMA NOTEBOOKS").collect()
session.sql("SELECT CURRENT_DATABASE(), CURRENT_SCHEMA()").show()
```
